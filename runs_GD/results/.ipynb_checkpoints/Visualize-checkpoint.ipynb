```python
import os
import webbrowser

# Configuration
PARAMS = {
    'noise': ['0.0', '0.125'],
    'cocktail': [str(i) for i in range(1, 49)],
    'drift': ['0.0', '0.0001', '0.001', '0.005']
}
BASE_PATH = "noise_{noise}/run_dr{drift}_ns{noise}_patt{cocktail}/composite.svg"

# Generate the HTML file
HTML_TEMPLATE = """
<!DOCTYPE html>
<html>
<head>
    <title>Figure Navigator</title>
    <style>
        body { font-family: Arial, sans-serif; margin: 20px; }
        .container { max-width: 1200px; margin: 0 auto; }
        .controls { display: flex; justify-content: space-between; margin-bottom: 20px; }
        .param-group { text-align: center; padding: 10px; border: 1px solid #ddd; border-radius: 5px; }
        button { padding: 5px 15px; margin: 0 5px; cursor: pointer; }
        #image-container { text-align: center; margin-top: 20px; min-height: 500px; }
        img { max-width: 100%; border: 1px solid #eee; }
        .error { color: red; font-weight: bold; }
    </style>
</head>
<body>
    <div class="container">
        <h1>Figure Navigator</h1>
        
        <div class="controls">
            <div class="param-group">
                <h3>Noise</h3>
                <div>
                    <button onclick="changeParam('noise', -1)">⬅</button>
                    <span id="noise-value">0.0</span>
                    <button onclick="changeParam('noise', 1)">➡</button>
                </div>
            </div>
            
            <div class="param-group">
                <h3>Cocktail</h3>
                <div>
                    <button onclick="changeParam('cocktail', -1)">⬅</button>
                    <span id="cocktail-value">1</span>
                    <button onclick="changeParam('cocktail', 1)">➡</button>
                </div>
            </div>
            
            <div class="param-group">
                <h3>Drift</h3>
                <div>
                    <button onclick="changeParam('drift', -1)">⬅</button>
                    <span id="drift-value">0.0</span>
                    <button onclick="changeParam('drift', 1)">➡</button>
                </div>
            </div>
        </div>
        
        <div id="image-container">
            <p>Select parameters to view the figure</p>
        </div>
    </div>

    <script>
        // Current parameter values
        const params = {
            noise: 0,
            cocktail: 0,
            drift: 0
        };
        
        // Parameter options
        const paramOptions = {
            noise: {{NOISE_VALUES}},
            cocktail: {{COCKTAIL_VALUES}},
            drift: {{DRIFT_VALUES}}
        };
        
        // Update the display
        function updateDisplay() {
            // Update parameter displays
            document.getElementById('noise-value').textContent = paramOptions.noise[params.noise];
            document.getElementById('cocktail-value').textContent = paramOptions.cocktail[params.cocktail];
            document.getElementById('drift-value').textContent = paramOptions.drift[params.drift];
            
            // Generate the image path
            const imagePath = `{{BASE_PATH}}`
                .replace('{noise}', paramOptions.noise[params.noise])
                .replace('{cocktail}', paramOptions.cocktail[params.cocktail])
                .replace('{drift}', paramOptions.drift[params.drift]);
            
            // Display the image or error
            const container = document.getElementById('image-container');
            container.innerHTML = `<img src="${imagePath}" onerror="this.onerror=null;showError('${imagePath}')">`;
        }
        
        // Show error message
        function showError(path) {
            const container = document.getElementById('image-container');
            container.innerHTML = `<p class="error">Image not found: ${path}</p>`;
        }
        
        // Change parameter value
        function changeParam(param, delta) {
            params[param] += delta;
            
            // Handle wrapping
            if (params[param] < 0) {
                params[param] = paramOptions[param].length - 1;
            } else if (params[param] >= paramOptions[param].length) {
                params[param] = 0;
            }
            
            updateDisplay();
        }
        
        // Initial display
        updateDisplay();
    </script>
</body>
</html>
"""

def generate_html():
    # Prepare the template
    html = HTML_TEMPLATE
    html = html.replace("{{NOISE_VALUES}}", str(PARAMS['noise']))
    html = html.replace("{{COCKTAIL_VALUES}}", str(PARAMS['cocktail']))
    html = html.replace("{{DRIFT_VALUES}}", str(PARAMS['drift']))
    html = html.replace("{{BASE_PATH}}", BASE_PATH)
    
    # Save to file
    with open("figure_navigator.html", "w") as f:
        f.write(html)
    
    print("HTML file generated: figure_navigator.html")
    webbrowser.open("figure_navigator.html")

if __name__ == "__main__":
    generate_html()
```

```python
import os
from IPython.display import display, clear_output, SVG
import ipywidgets as widgets

# Define your parameter sets
param1_values = ['0.0', '0.125']  # noise
param2_values = [str(i) for i in range(1, 47)]  # cocktail
param3_values = ['0.0', '0.0001', '0.001', '0.005']  # drift

# Initialize current indices
current_indices = {'param1': 0, 'param2': 0, 'param3': 0}

# Create output area
output = widgets.Output()

# Create buttons with better styling
button_style = {'button_color': 'lightgray', 'font_weight': 'bold'}
button_layout = widgets.Layout(width='150px', margin='5px')

prev_param1 = widgets.Button(description='⬅ Noise', style=button_style, layout=button_layout)
next_param1 = widgets.Button(description='Noise ➡', style=button_style, layout=button_layout)
prev_param2 = widgets.Button(description='⬅ Cocktail', style=button_style, layout=button_layout)
next_param2 = widgets.Button(description='Cocktail ➡', style=button_style, layout=button_layout)
prev_param3 = widgets.Button(description='⬅ Drift', style=button_style, layout=button_layout)
next_param3 = widgets.Button(description='Drift ➡', style=button_style, layout=button_layout)

# Status label
status_label = widgets.Label(value="Ready...", layout=widgets.Layout(margin='10px 0'))

def update_display():
    """Update the display with current parameters"""
    with output:
        clear_output(wait=True)
        
        p1 = param1_values[current_indices['param1']]
        p2 = param2_values[current_indices['param2']]
        p3 = param3_values[current_indices['param3']]
        
        filename = f'results/run_dr{p3}_ns{p1}_patt{p2}/composite.svg'
        filepath = os.path.join('./', filename)
        
        status_label.value = f"Noise: {p1} | Cocktail: {p2} | Drift: {p3}"
        
        try:
            if os.path.exists(filepath):
                display(SVG(filename=filepath))
            else:
                print(f"File not found: {filename}")
        except Exception as e:
            print(f"Error displaying {filename}: {str(e)}")

def cycle_param(param, direction):
    """Cycle through parameter values"""
    params = {
        'param1': param1_values,
        'param2': param2_values,
        'param3': param3_values
    }
    
    current_idx = current_indices[param] + direction
    max_idx = len(params[param]) - 1
    
    # Handle wrapping
    if current_idx > max_idx:
        current_idx = 0
    elif current_idx < 0:
        current_idx = max_idx
    
    current_indices[param] = current_idx
    update_display()

# Assign button actions
prev_param1.on_click(lambda _: cycle_param('param1', -1))
next_param1.on_click(lambda _: cycle_param('param1', 1))
prev_param2.on_click(lambda _: cycle_param('param2', -1))
next_param2.on_click(lambda _: cycle_param('param2', 1))
prev_param3.on_click(lambda _: cycle_param('param3', -1))
next_param3.on_click(lambda _: cycle_param('param3', 1))

# Layout the interface
controls = widgets.VBox([
    widgets.HBox([prev_param1, next_param1]),
    widgets.HBox([prev_param2, next_param2]),
    widgets.HBox([prev_param3, next_param3]),
    status_label
], layout=widgets.Layout(margin='20px 0'))

ui = widgets.VBox([
    widgets.HTML("<h2 style='text-align:center'>Figure Navigator</h2>"),
    controls,
    output
], layout=widgets.Layout(width='100%'))

display(ui)
update_display()
```
